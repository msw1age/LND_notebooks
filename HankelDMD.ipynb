```python
import numpy as np
from pydmd import HankelDMD, HODMD, BOPDMD,DMD
from MrOptDMD import MrOptDMD
from pydmd.utils import compute_svd
import matplotlib.pylab as plt
plt.rcParams['figure.dpi']=400
np.set_printoptions(suppress=True,precision=6,linewidth=100000) # Improve print statements
```

```python
RSE = lambda X,Y: np.sum((Y-np.real(X))**2)/np.sum((Y-np.mean(Y))**2)
# Model reconstruction = X, true data = Y
```

```python
SV = np.load('../Mouse Data/svdTemporalComponents_corr.npy')[:2000,:50].T
SV_norm = np.linalg.norm(SV)
SV_normalized = SV / SV_norm
```

```python
dmd = DMD(24)
hdmd = HankelDMD(24,reconstruction_method='mean',d=950)
optdmd = BOPDMD(24)
mroptdmd = MrOptDMD([BOPDMD(10),BOPDMD(6),BOPDMD(8)],max_level=2,max_cycles=np.inf)
```

```python
dmd.fit(SV_normalized[:,:1000])
hdmd.fit(SV_normalized[:,:1000])
mroptdmd.fit(SV_normalized[:,:1000],np.arange(1000))
optdmd.fit(SV_normalized[:,:1000],np.arange(1000))
```

```python
print("DMD RSE: " + str(RSE(dmd.reconstructed_data,SV_normalized[:,:1000])))
print("Hankel DMD RSE: " + str(RSE(hdmd.reconstructed_data,SV_normalized[:,:1000])))
print("MrOptDMD RSE: " + str(RSE(mroptdmd.reconstructed_data,SV_normalized[:,:1000])))
print("OptDMD RSE: " + str(RSE(optdmd.reconstructed_data,SV_normalized[:,:1000])))
```

```python
# Takes the hankel matrix and recovers the original matrix
# I think it does this by averaging across skew-diagonals
# This is taken from the pydmd source
def average_timepoints(data):
    rec = data
        
    reconstructed_snapshots = np.full((time_instants, hdmd.d, space_dim), np.nan, dtype=rec.dtype)
    c_idxes = (np.array(range(hdmd.d))[:, None].repeat(2, axis=1)[None, :].repeat(rec.shape[1], axis=0))
    c_idxes[:, :, 0] += np.array(range(rec.shape[1]))[:, None]    
    reconstructed_snapshots[c_idxes[:, :, 0], c_idxes[:, :, 1]] = np.array(
            np.swapaxes(np.split(rec.T, hdmd.d, axis=1), 0, 1)
        )
    rec2 = reconstructed_snapshots
    rec2 = np.ma.array(rec2, mask=np.isnan(rec2))
    result = np.mean(rec2, axis=1).T
    result = result.filled(fill_value=0)
    return result
```

```python
# This computes the one step update. Here we do it on the training data and it works very well.
forecast = hdmd._sub_dmd.predict(hdmd._pseudo_hankel_matrix(SV_normalized[:,:1000]))
forecast = average_timepoints(forecast)
print("Train RSE: " + str(RSE(forecast,SV_normalized[:,1:1001])))
```

```python
# Now we compute the one step update on validation data on a window
# of data including 50 new timepoints. This fails upon getting to the data the model wasn't trained on.
# Maybe there is a better way to do forecasting.
forecast2 = hdmd._sub_dmd.predict(hdmd._pseudo_hankel_matrix(SV_normalized[:,50:1050]))
forecast2 = average_timepoints(forecast2)
print("Val RSE: " + str(RSE(forecast2[:,-50:],SV_normalized[:,1001:1051])))
```

```python
# Plot comparing forecast2 above to true data. The blue line is where the training data ends.
fig, ax = plt.subplots(2,1,sharex=True,sharey=True)
for i in range(10):
    ax[0].plot(forecast2[i,:])
    ax[0].axvline(950)
    ax[1].plot(SV_normalized[i,51:1051])
```

```python
# Strangely, hdmd's dynamics function produce modes that decay to zero almost immediately. I think this maybe not a great
# implementation of this function. The dynamics function just computes the eigenvalue powers and multiplies that by the amplitudes
# (which in this case are the projection of the initial condition onto the eigenvector basis). However, the eigenvalues 
# are fit based on the timescale that the exact DMD sees, which in this case is just 51 timepoints (since the Hankel matrix
# version of the data is 47500x51), so the time scale is off when we try to get the dynamics using 1000 timepoints.

# The HankelDMD reconstruction does not use hdmd.dynamics, which is 24x1000, but rather hdmd._sub_dmd.dynamics,
# which is 24x51.

for i in range(0,24,2):
    plt.plot(np.real(hdmd.dynamics[i]))
    plt.figure()
```
